# 🧠 ConfLab Dataset Preprocessing: JSON to DataFrame

This notebook parses raw ConfLab skeleton annotation JSON files into a structured `pandas.DataFrame`. Each row corresponds to an individual agent's keypoints and occlusion status in a given frame. The resulting DataFrame can be used for downstream analysis such as visibility scoring, camera selection, and sliding window feature extraction.

**Contents:**
- Load and parse JSON annotation files
- Extract keypoints and occlusion masks
- Structure data per video, camera, frame, and agent

```python
#imports
import numpy as np
import pandas as pd
import json
import os
from tqdm import tqdm
```

```python
def parse_conflab_skeleton_json(data, video_id, camera_id):
    """
    Parses ConfLab skeleton annotation JSON into a structured DataFrame.

    Args:
        json_data (dict): Parsed JSON object from a ConfLab annotation file.
        video_id (str or int): Identifier for the video session.
        camera_id (str or int): Identifier for the camera.

    Returns:
        pd.DataFrame: Each row is a single agent-frame observation.
                      Columns: video_id, camera_id, frame_id, agent_id, keypoints (Nx2), occluded (N,)
    """

    rows = []
    skeletons_by_frame = data["annotations"]["skeletons"]
    for frame_index, frame_skeletons in tqdm(enumerate(skeletons_by_frame), desc=f"Parsing skeleton annotations for {video_id}_{camera_id}"):
        for agent_id_str, skeleton_data in frame_skeletons.items():
            agent_id = int(agent_id_str)
            num_kp = skeleton_data["num_keypoints"]
            keypoints = np.array(skeleton_data["keypoints"]).reshape(-1, 2)
            occluded = np.array(skeleton_data["occluded"])
            frame_id = frame_index

            rows.append({
                "video_id": video_id,
                "camera_id": camera_id,
                "frame_id": frame_id,
                "agent_id": agent_id,
                "num_kp": num_kp,
                "keypoints": keypoints,
                "occluded": occluded
            })
    return pd.DataFrame(rows)

```

```python
def extract_ids_from_filename(filename):
    base = os.path.basename(filename).replace(".json", "")
    parts = base.split("_")
    video_id = parts[1] + "_" + parts[2]  # e.g., "conf01"
    camera_id = parts[0]
    return video_id, camera_id

def parse_all_jsons_in_folder(folder_path):
    all_dfs = []
    for f_id, file in enumerate (os.listdir(folder_path)):
        print(f"Parsing {file}; {f_id}/32")
        if file.endswith(".json"):
            json_path = os.path.join(folder_path, file)
            with open(json_path) as f:
                data = json.load(f)

            video_id, camera_id = extract_ids_from_filename(file)
            df = parse_conflab_skeleton_json(data, video_id, camera_id)
            all_dfs.append(df)

    full_df = pd.concat(all_dfs, ignore_index=True)
    return full_df
```

## 😂 Adding Laughter Annotations to Agents per Frame

The following code blocks enrich our existing pose-tracking DataFrame by incorporating **laughter annotations**.
### What This Will Do:
- Load or read external laughter annotation files for each modality.
- Align annotations with the agent and frame IDs already present in the pose DataFrame.
- Add new columns for each modality with binary label for each `(agent_id, frame_id)` pair.

```python
def csv_to_pd(folder_name, file, index):
    full_path = os.path.join(folder_name, file)
    fp_a = full_path + f"_ann{index}.csv"
    df = pd.read_csv(fp_a)
    df = df.rename(columns={col: f"agent {col} ann{index}" for col in df.columns})
    return df

def generate_big_annotations(folder_name, file):
    csv_1 = csv_to_pd(folder_name, file, 1)
    csv_2 = csv_to_pd(folder_name, file, 2)
    csv_3 = csv_to_pd(folder_name, file, 3)

    agents_1 = [col.split()[1] for col in csv_1.columns]
    agents_2 = [col.split()[1] for col in csv_2.columns]
    agents_3 = [col.split()[1] for col in csv_3.columns]
    all_agents = sorted(set(agents_1) | set(agents_2) | set(agents_3), key=int)

    for agent in all_agents:
        csv_1[f"agent {agent} ann1"] = csv_1.get(f"agent {agent} ann1", None)
        csv_2[f"agent {agent} ann2"] = csv_2.get(f"agent {agent} ann2", None)
        csv_3[f"agent {agent} ann3"] = csv_3.get(f"agent {agent} ann3", None)

# Ensure consistent column order
    csv_1 = csv_1.sort_index(axis=1)
    csv_2 = csv_2.sort_index(axis=1)
    csv_3 = csv_3.sort_index(axis=1)


    interleaved_cols = [
    f"agent {agent} ann{i}"
    for agent in all_agents
    for i in range(1, 4)  # ann1, ann2, ann3
    ]


    combined_df = pd.concat([csv_1, csv_2, csv_3], axis=1)
    combined_df["frame_id"] = combined_df.index
    combined_df["video_id"] = file
    combined_df = combined_df[["frame_id", "video_id"] + interleaved_cols]


    return combined_df

def gen_pd_for_modality(folder_name, test_videos = None):
    if test_videos is None:
        pose_annotated_segments = ["vid2_seg8", "vid2_seg9", "vid3_seg1", "vid3_seg2", "vid3_seg3", "vid3_seg4", "vid3_seg5", "vid3_seg6"]
    else:
        pose_annotated_segments = test_videos
    al_dfs = []
    for file in pose_annotated_segments:
       df = generate_big_annotations(folder_name, file)
       al_dfs.append(df)
    full_df = pd.concat(al_dfs, ignore_index=True)
    return full_df
```

## Combining the dataframes together
Make so there is one dataframe with all the data:
- Pose estimation coordinates
- Laughter indication for each modality (majority rule)

```python
df = pd.read_pickle("dataframes\\conflab_skeleton_data.pkl")
```

```python
def majority_vote(vals):
    filtered = [v for v in vals if pd.notna(v)]
    if not filtered:
        return None
    return max(set(filtered), key=filtered.count)

def assign_majority_annotations(pose_df, ann_df, modality):
    # Make sure merge keys are aligned
    key_cols = ["video_id", "frame_id"]
    ann_df = ann_df.set_index(key_cols)
    agents_ids = []
    cams_ids = []
    video_ids = []
    nums = []

    results = []
    for i, row in pose_df.iterrows():
        video_id = row["video_id"]
        frame_id = row["frame_id"]
        agent_id = row["agent_id"]
        camera_id = row["camera_id"]

        try:
            ann_row = ann_df.loc[(video_id, frame_id)]
        except KeyError:
            results.append(None)
            continue

        agent_cols = [col for col in ann_row.index if col.startswith(f"agent {agent_id} ann")]
        agent_votes = [ann_row[col] for col in agent_cols if col in ann_row]

        final_vote = majority_vote(agent_votes)
        if final_vote is None:
            agents_ids.append(agent_id)
            cams_ids.append(camera_id)
            video_ids.append(video_id)
            nums.append(i)
        results.append(final_vote)
    pose_df[f"is_laughing_{modality}"] = results
    print(f"{modality} nan: {pose_df[f'is_laughing_{modality}'].isna().sum()}")
    print(f"Agent IDs: {np.unique(agents_ids)}")
    print(f"Video IDs: {np.unique(video_ids)}")
    print(f"Camera IDs: {np.unique(cams_ids)}")
    return pose_df, nums

```

```python
final_df_na,frames = assign_majority_annotations(df, df_no_audio, "no_audio")
final_df_oa,frames = assign_majority_annotations(final_df_na, df_only_audio, "only_audio")
final_df_wa,frames = assign_majority_annotations(final_df_oa, df_with_audio, "with_audio")
display(final_df_wa)
```
